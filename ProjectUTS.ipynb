```python
class CoffeeQueue:
    def __init__(self):
        self.antrian = []

    def tampil_menu(self):
        menu = [
            "Matcha Latte",
            "Vanilla Latte",
            "Caramel Latte",
            "Kopi Gula Aren",
            "Cappuccino",
            "Americano",
            "Mochachino",
            "Frappuccino",
            "Espresso",
            "Caramel Macchiato"
        ]

        print("\n MENU MINUMAN")
        print("-" * 30)
        for i, item in enumerate(menu, 1):
            print(f"{i}. {item}")
        print("-" * 30)

        return menu

    # ENQUEUE
    def tambah(self, nama, minum):
        data = {
            "nama": nama,
            "menu": minum
        }
        self.antrian.append(data)

        print(f"\n Nama: {nama}")
        print(f"Pesanan: {minum}")
        print(f"Posisi: {len(self.antrian)}")

    # DEQUEUE
    def layani(self):
        if len(self.antrian) == 0:
            print("\nAntrian kosong")
        else:
            p = self.antrian.pop(0)
            print("\nMelayani: ")
            print(f"{p['nama']}")
            print(f"{p['menu']}")

    # PEEK
    def depan(self):
        if len(self.antrian) == 0:
            print("\nAntrian kosong")
        else:
            p = self.antrian[0]
            print("\nAntrian paling depan:")
            print(f"{p['nama']} - {p['menu']}")

    # DISPLAY
    def tampilkan(self):
        if len(self.antrian) == 0:
            print("\nAntrian kosong")
        else:
            print("\nList antrian")
            print("-" * 35)
            for i, p in enumerate(self.antrian, 1):
                print(f"{i}. {p['nama']} ({p['menu']})")
            print("-" * 35)


# Main
kopi = CoffeeQueue()

while True:
    print("\n" + "="*43)
    print("      SISTEM ANTRIAN COFFEE SHOP")
    print("="*43)
    print("1. Tambah pelanggan")
    print("2. Layani pelanggan")
    print("3. Lihat antrian paling depan")
    print("4. Tampilkan semua antrian")
    print("5. Keluar")

    pilih = input("Pilih menu: ")

    if pilih == "1":
        print("\n=== Input Pelanggan ===")
        nama = input("Nama: ")

        menu = kopi.tampil_menu()
        pilihmenu = input("Pilih nomor menu: ")

        if pilihmenu.isdigit():
            pilihmenu = int(pilihmenu)

            if 1 <= pilihmenu <= len(menu):
                minum = menu[pilihmenu - 1]
            else:
                minum = "Menu tidak tersedia"
        else:
            minum = "Menu tidak tersedia"

        kopi.tambah(nama, minum)

    elif pilih == "2":
        kopi.layani()

    elif pilih == "3":
        kopi.depan()

    elif pilih == "4":
        kopi.tampilkan()

    elif pilih == "5":
        print("\nProgram selesai")
        break

    else:
        print("\nPilihan tidak ada!")
```
