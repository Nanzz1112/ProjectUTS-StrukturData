```python
antrian = []

menu = [
    "Matcha Latte",
    "Vanilla Latte",
    "Caramel Latte",
    "Kopi Gula Aren",
    "Cappuccino",
    "Americano",
    "Mochachino",
    "Frappuccino",
    "Espresso",
    "Caramel Macchiato",
    "Hazelnut Coffee",
]

while True:
    print("\n" + "—"*43)
    print("        SISTEM ANTRIAN COFFEE SHOP")
    print("—"*43)
    print("1. Tambah pelanggan")
    print("2. Layani pelanggan")
    print("3. Lihat antrian paling depan")
    print("4. Tampilkan semua antrian")
    print("5. Keluar")
    print("—"*43)
    pilihan = input("Pilih menu (1-5): ")

    if pilihan == "1":
        print("\n ===== Tambah Pelanggan =====")
        nama = input("Nama pelanggan: ")

        print("\n DAFTAR MENU MINUMAN:")

        for i in range(len(menu)):
            print(f"{i+1}. {menu [i]}")

        try:
            pilih_menu = int(input("Pilih nomor menu (1-11): "))

            if 1 <= pilih_menu <= len(menu):
                minuman = menu [pilih_menu - 1]

                data = nama + " = " + minuman
                antrian.append(data)
                print(f" {nama} Memilih Menu {pilih_menu} ({minuman})")

            else:
                print("\n Nomor menu tidak tersedia!")

        except:
            print("\n Input harus berupa angka!")
```
